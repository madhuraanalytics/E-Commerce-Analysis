# 22. Create age groups:18-25,26-35,36-45,46+


```python
def age_group(age):
    if 18<= age <=25:
        return "18-25"
    elif 26<= age <=35:
        return "26-35"
    elif 36<= age <=45:
        return "36-45"
    else: 
        return "46+"

df["age_group"]=df["age"].apply(age_group)
print(df[["age_group","age"]])
    
```
