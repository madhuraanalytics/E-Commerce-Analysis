# Q16. Find monthly growth in orders.


```python
monthly_growth=(df["order_date"].dt.month_name().value_counts().sort_values())
print(monthly_growth)
```

```python
monthly_orders = (
    df.groupby(df["order_date"].dt.to_period("M"))
      .size()
)

monthly_growth = monthly_orders.pct_change() * 100

print(monthly_growth)
```

# Q17. Create rolling 30-day order trend.


```python
var=df.groupby("order_date").size()
print(var)
```

```python
daily_order=var.rolling(window=31).sum()
print(daily_order)
```
