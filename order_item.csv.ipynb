# Q5. Create revenue column


```python
df["revenue"]=df["quantity"]*df["unit_price"]
```

```python
df
```

# Q6. Create discount amount.


```python
df["discount_amt"]=(round(df["discount_pct"]*df["revenue"]/100,2))
#df["discount_amt"]=(df["discount_pct"]*df["revenue"]/100).round(2)
```

```python
df["discount_pct"].head(10)
```

```python
df
```

# Q7. Create final revenue.

```python
df["total_revenue"]=df["revenue"]+df["discount_amt"]
```

```python
df
```

# Q8. Find top 10 revenue-generating products.


```python
df.groupby("product_id")["total_revenue"].sum().sort_values(ascending=False).head(10)
```

# Q9. Find highest discount product.


```python
df.groupby("product_id")["discount_amt"].max().sort_values(ascending=False)
```
